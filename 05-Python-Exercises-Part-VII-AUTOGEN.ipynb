1. **Compounding Interest Formula**:
   - The cumulative amount \( A \) for a given month is calculated as:<br>

     $A = P \times \left(1 + \frac{r}{12}\right)^n$

     where:
     - $P$ is the principal amount.
     - $r$ is the annual interest rate (as a decimal).
     - $n$ is the number of months elapsed.

2. **Earnings**:
   - Earnings for a given month are $ A - P $.
