### Squares

**Objective**: Create a class **square_manager** to manage **squares** with sides parallel to the x and y axes.

The class **square** should have the following attributes:  
- **Bottom-left vertex** (`bs`): A tuple representing its coordinates in the plane.  
- **Side length** (`side`): An integer representing the length of the side of the square.

and the following methods:
1. `get_area(self)`: Returns the area of the square.
2. `get_perimeter(self)`: Returns the perimeter of the square.
3. `__str__(self)`: Returns a string representing the square in the format `"[bs, side, area, perimeter]"`.

**Note:** you cannot have squares with negative side length

The class **square_manager** should have a list `squares` of squares as an attribute and the following methods:
1. `add_square(self, bs, side)`: Adds a new square to the list. `bs` is the bottom-left vertex, and `side` is the side length.
2. `delete_squares_area(self, area)`: Removes all squares with an area less than a given input value.
3. `__str__(self)`: Returns a multiline string, where each row represents a square. Each square is represented with:
     - Bottom-left vertex `bs` and side length.
     - Its Area.
     - Its Perimeter.

---

### Examples for a Square:
- **Bottom-left vertex (`bs`)**: (2, 3)  
- **Side length (`side`)**: `5`  

### Results of Methods:
1. `get_area()`: `25`
2. `get_perimeter()`: `20`
3. `__str__()`: `"[(2,3), 5, 25, 20]"`

```python
# Solution goes here
class Square():
    def __init__(self, bs, side):
        if side < 0:
            raise(ValueError("Invalid input"))
        self.bs = bs
        self.side = side
    def get_area(self):
        return self.side**2
    def get_perimeter(self):
        return self.side*4
    def __str__(self):
        return f"[{self.bs}, {self.side}, {self.get_area()}, {self.get_perimeter()}]"

class SquareManager():
    def __init__(self):
        self.squares = []
    def add_square(self, bs, side):
        self.squares.append(Square(bs, side))
    def delete_squares_area(self, area):
        result = []
        for square in self.squares:
            if square.get_area() >= area:
                result.append(square)
        self.squares = result
    def __str__(self):
        if len(self.squares) == 0:
            return ""
        result = f"[{self.squares[0].bs}, {self.squares[0].side}, {self.squares[0].get_area()}, {self.squares[0].get_perimeter()}]"
        for i in range(1, len(self.squares)):
            result += f"\n[{self.squares[i].bs}, {self.squares[i].side}, {self.squares[i].get_area()}, {self.squares[i].get_perimeter()}]"
        return result
```

## Exercise 2


You are given a dataset named [**`iris.csv`**](iris.csv), which contains measurements in centimeters of iris flowers (sepal length, sepal width, petal length, petal width) along with their species labels.

| sepal_length | sepal_width | petal_length | petal_width | species       |
|--------------|-------------|--------------|-------------|---------------|
| 5.1          | 3.5         | 1.4          | 0.2         | Iris-setosa   |
| 4.9          | 3.0         | 1.4          | 0.2         | Iris-setosa   |
| 7.0          | 3.2         | 4.7          | 1.4         | Iris-versicolor |
| 6.3          | 3.3         | 6.0          | 2.5         | Iris-virginica |

Your task is to build a **binary classifier** using a **Gaussian Naive Bayes classifier**.

#### **Instructions**

1. Load the dataset into a pandas DataFrame.
2. Transform the `species` column into a new binary label and store it in a new column named **`label`**:
   - Label as **1** if `species == 'Iris-setosa'`
   - Label as **0** otherwise
3. Select these numeric columns as features, store them in **`X`**, and the binary labels in **`y`**:
   - `sepal_length`
   - `sepal_width`
   - `petal_length`
   - `petal_width`
4. Split the data into training and testing sets using the following variables:
   * `X_train`, `X_test`, `y_train`, `y_test`
   * Use `test_size=0.20` and `stratify=y` in `train_test_split`.

6. Train a **Gaussian Naive Bayes classifier** (`GaussianNB()`) using the training data and store the model in a variable named **`model`**.

7. Make predictions on the test set and store them in a variable named **`y_pred`**.

8. Calculate:
   * The **accuracy score** (store it in a variable named `accuracy`)
   * The **confusion matrix** (store it in a variable named `conf_matrix`)


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix
import numpy as np
# Solution goes here
df = pd.read_csv("iris.csv")
df["label"] = df["species"].apply(lambda x: 1 if x == "Iris-setosa" else 0)
X = df.select_dtypes(include = ["float64"])
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, stratify=y)

model = GaussianNB()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
```
