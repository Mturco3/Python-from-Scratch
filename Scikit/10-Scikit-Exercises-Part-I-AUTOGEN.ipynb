## Exercise

You are given a dataset called [**`sentiment_analisys.csv`**](sentiment_analisys.csv), which contains a collection of tweets in a column named **`Tweet`** and their corresponding sentiment scores in a column named **`Avg`**.

Your task is to build a **binary sentiment classifier** using a **Support Vector Machine (SVM)** with a linear kernel.

#### **Instructions**

1. Load the dataset into a pandas DataFrame.
2. Convert all tweets in the column **`Tweet`** to lowercase.
3. Transform the text data into a **bag-of-words** representation using `CountVectorizer` from scikit-learn, removing English stop words.
4. Convert the sentiment column (`Avg`) into binary labels as follows:

   * Label as **1** if `Avg <= -1` (negative sentiment)
   * Label as **0** otherwise.
5. Split the data into training and testing sets using the following variables:

   * `X_train`, `X_test`, `y_train`, `y_test`
   * Use `test_size=0.25` and `stratify=df['Avg']` in `train_test_split`.
6. Train a **linear SVM classifier** (`LinearSVC()`) using the training data.
7. Make predictions on the test set and store them in a variable named **`y_pred`**.
8. Calculate:

   * The **accuracy score** (store it in a variable named `accuracy`)
   * The **confusion matrix** (store it in a variable named `conf_matrix`)

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, confusion_matrix
import numpy as np
# Solution goes here
df = pd.read_csv("sentiment_analisys.csv")
df["Tweet"] = df["Tweet"].str.lower()
df["Avg"] = df["Avg"].apply(lambda x: 1 if x <= -1 else 0)

vectorizer = CountVectorizer(stop_words= "english")
X = vectorizer.fit_transform(df["Tweet"])

X_train, X_test, y_train, y_test = train_test_split(X, list(df["Avg"]), test_size= 0.25, stratify=df["Avg"])
svm_classifier = LinearSVC()
svm_classifier.fit(X_train, y_train)
y_pred = svm_classifier.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)

```
