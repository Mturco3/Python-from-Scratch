# Comprehensions

Comprehensions construct collections by transforming and optionally filtering values from an iterable. List, set, and dictionary comprehensions evaluate eagerly; generator expressions produce values lazily.

The examples use Python 3 without external packages or datasets.

## List comprehensions

The form `[expression for item in iterable]` evaluates an expression for each item and collects the results in a new list. The list follows iteration order and retains duplicate results. The source collection is not modified by these examples.

````python
numbers = [1, 2, 3, 4, 5]
squares_loop = []
for number in numbers:
    squares_loop.append(number ** 2)

squares = [number ** 2 for number in numbers]
print(squares)  # [1, 4, 9, 16, 25]
print(squares == squares_loop)  # True
````

## Filtering

A trailing `if` includes only items satisfying a condition:

```python
[expression for item in iterable if condition]
```

The condition is evaluated before the expression. Filtering can leave fewer results than input items, including an empty list.

````python
words = ["I", "am", "Michele"]
long_words_loop = []
for word in words:
    if len(word) > 3:
        long_words_loop.append(word)

long_words = [word for word in words if len(word) > 3]
print(long_words)  # ['Michele']
print(long_words == long_words_loop)  # True
print([number ** 2 for number in numbers if number % 2 == 0])  # [4, 16]
print([word for word in words if len(word) > 20])  # []
````

## Conditional expressions

A conditional expression chooses the value produced for an item:

```python
[value_if_true if condition else value_if_false for item in iterable]
```

Unlike a trailing filter, it does not omit an item. A conditional expression and a filter can appear together; the filter first determines which items reach the expression.

````python
labels = ["even" if number % 2 == 0 else "odd" for number in numbers]
print(labels)  # ['odd', 'even', 'odd', 'even', 'odd']

values = [-2, -1, 0, 1, 2]
signs = ["positive" if value > 0 else "negative" for value in values if value != 0]
print(signs)  # ['negative', 'negative', 'positive', 'positive']
````

## Set comprehensions

The form `{expression for item in iterable}` collects distinct results in a set. Set elements must be hashable. A set has no guaranteed iteration order; sorting the result produces a predictable display in this example.

Empty braces `{}` create a dictionary. An empty set is created with `set()`.

````python
names = [" Python ", "python", "PANDAS", " pandas "]
normalized_names = {name.strip().lower() for name in names}
print(sorted(normalized_names))  # ['pandas', 'python']
````

## Dictionary comprehensions

The form `{key_expression: value_expression for item in iterable}` constructs a dictionary. Keys must be hashable. Repeated keys retain the last assigned value; values do not need to be unique.

````python
lengths = {word: len(word) for word in words}
print(lengths)  # {'I': 1, 'am': 2, 'Michele': 7}

pairs = [("python", 1), ("pandas", 2), ("python", 3)]
latest = {name: value for name, value in pairs}
print(latest)  # {'python': 3, 'pandas': 2}
````

## Multiple clauses and nested comprehensions

Multiple `for` clauses follow the same left-to-right order as nested loops. In `[value for row in matrix for value in row]`, the row is selected before its values are traversed. This produces a flat list.

A comprehension inside another comprehension preserves a nested structure.

````python
matrix = [[1, 2, 3], [4, 5, 6]]
flattened = [value for row in matrix for value in row]
scaled_rows = [[value * 10 for value in row] for row in matrix]
coordinates = [(row, column) for row in range(2) for column in range(3)]

print(flattened)  # [1, 2, 3, 4, 5, 6]
print(scaled_rows)  # [[10, 20, 30], [40, 50, 60]]
print(coordinates)  # [(0, 0), (0, 1), (0, 2), (1, 0), (1, 1), (1, 2)]
````

## Eager collections and lazy generator expressions

A generator expression uses parentheses instead of list brackets. It returns an iterator whose results are computed on demand. Consuming the iterator advances its state; an exhausted generator produces no further values.

The leftmost iterable expression is evaluated when the generator expression is created. The result expression is evaluated during consumption. Laziness avoids storing all computed results at once, but does not remove the memory occupied by existing source data. Converting a generator to a list materializes its remaining results.

Related concepts: [iterators and generators](../Iterators_Generators/iterators_and_generators.ipynb).

````python
eager_squares = [number ** 2 for number in range(4)]
lazy_squares = (number ** 2 for number in range(4))

print(eager_squares)  # [0, 1, 4, 9]
print(next(lazy_squares))  # 0
print(list(lazy_squares))  # [1, 4, 9]
print(list(lazy_squares))  # []
print(sum(number ** 2 for number in range(4)))  # 14
````

## Readability and side effects

Comprehensions fit short transformations and filters. Explicit loops provide clearer structure for several processing steps, exception handling, or side effects such as writing files. A comprehension used only for side effects also constructs an unnecessary result collection.

List, set, and dictionary comprehensions store their results in memory. Generator expressions fit incremental consumers such as `sum()`; lists fit indexing and repeated traversal.
