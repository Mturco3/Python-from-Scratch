# Regular Expressions

The `re` module provides support for matching text against patterns using regular expressions. It is part of the Python standard library and does not require installation.

Documentation: [https://docs.python.org/3/library/re.html](https://docs.python.org/3/library/re.html)

```python
import re
```

## Basic patterns

We need to specify a pattern and use `finditer` to find matches in a string.

| Symbol | Meaning |
|---|---|
| `.` | Any character |
| `\w` | Any word character `[A-Za-z0-9_]` |
| `\W` | Any non-word character |
| `\d` | Any digit |
| `\D` | Not a digit |
| `\s` | Whitespace (space, tab, newline) |
| `\S` | Not a whitespace |
| `\b` | Word boundary |
| `\B` | Not a word boundary |
| `^` | Beginning of a string |
| `$` | End of a string |

```python
simple_pattern = re.compile(r"miche\.")
simple_string = "miche.turco.miche."
matches = simple_pattern.finditer(simple_string)
for match in matches:
    print(match)
```

## Quantifiers

| Symbol | Meaning |
|---|---|
| `*` | 0 or more |
| `+` | 1 or more |
| `?` | 0 or 1 |
| `{3}` | Exact number |
| `{3,5}` | Range of numbers |

### Pattern: 3 letters, 3-5 digits, 1 symbol

If `^` is inside square brackets, it negates the set.

```python
pattern1 = re.compile("^[A-Za-z]{3}[0-9]{3,5}[^A-Za-z0-9]{1}$")
print(pattern1.search("ABC3456!"))
print(pattern1.search("AB3456!"))
```

### Pattern: exactly 10 characters

```python
pattern2 = re.compile("^.{10}$")
print(pattern2.search("1234567890"))
```

## Practical patterns

### Email address

Some characters need to be escaped with a backslash: `. ^ ( ) $ * + [ ] { } ? \ |`

```python
mail_pattern = re.compile(r"^[A-Za-z0-9\.\-_]+@{1}[A-Za-z0-9\.\-_]+\.{1}[A-Za-z]{2,3}$")
print(mail_pattern.search("mymail.-2@gmail.it"))
```

### Phone numbers

```python
# This pattern accepts any separator because of the dot
phone_pattern_simple = re.compile(r"\d\d\d.\d\d\d.\d\d\d\d")

# This pattern requires specific separators
phone_pattern = re.compile(r"\d{3}[-_.]+\d{3}[._-]+\d{4}")

phone_number = "389_453_7700 320.---.842..2413"

print("Simple pattern:")
for match in phone_pattern_simple.finditer(phone_number):
    print(match)

print("\nStrict pattern:")
for match in phone_pattern.finditer(phone_number):
    print(match)
```

## Groups

With brackets `()` we can specify groups in the pattern. With `|` we can specify alternatives within a group.

```python
names = """
Mr. Turco
Mrs Rossi
Ms Giusi
Mrs. Blu
"""

formal_name = re.compile(r"[Mm][rs]?[s]?\.?\s[A-Za-z]+")
formal_name_with_groups = re.compile(r"((Mr|Ms|Mrs)\.?\s[A-Za-z]+)")

for match in formal_name.finditer(names):
    print(match)
```

### URL matching and substitution

The entire match is group 0. All other groups can be accessed with `.group(n)`.
The `sub` method substitutes each match with the selected group.

```python
urls = """https://www.google.com
http://coremys.com
https://youtube.com
https://www.nasa.gov"""

urls_pattern = re.compile(r"https?://(www\.)?(([A-Za-z]+)(.com|.gov))")

for match in urls_pattern.finditer(urls):
    print(match.group(0))

# Substitute each URL with just the domain
adjusted_urls = urls_pattern.sub(r"\2", urls)
print(adjusted_urls)
```

## Methods comparison

- `finditer`: returns match objects with position info
- `findall`: returns only the matched groups as strings
- `match`: matches only at the beginning of a string
- `search`: matches anywhere in the string

```python
# findall returns groups as tuples
matching_names = formal_name_with_groups.findall(names)
print(matching_names)
```

```python
sentence = "Start with a sentence"

# match only checks the beginning
print(re.compile(r"a").match(sentence))

# search finds matches anywhere
print(re.compile(r"a").search(sentence))
```

## Flags

Use `re.I` (or `re.IGNORECASE`) for case-insensitive matching.

```python
matches = re.compile(r"start", re.I).search(sentence)
print(matches)
```
