# Lists

Lists are ordered, mutable sequences. They can contain values of one type or a mixture of types, including other lists. Square brackets create list literals; `list()` constructs a list from an iterable or creates an empty list.

## Indexing, slicing, and traversal

Indices start at zero and extend to `len(values) - 1`. Negative indices count from the end. Indexing selects one element; slicing creates a new outer list containing references to the selected elements.

Direct iteration visits the elements. `range(len(values))` provides indices, while `enumerate()` supplies an index and a value together. Membership with `in` checks whether an equal element is present.

```python
values = [10, 20, 30]
print(values[0], values[-1])  # 10 30
print(values[1:])            # [20, 30]
print(20 in values)          # True
for index, value in enumerate(values):
    print(index, value)

mixed = ["a", 1, ["Hello", "Student"]]
print(mixed[-1][0])          # Hello
```

## Mutation and construction

Assigning to an index replaces an element. Assigning to a slice can replace several elements and change the length. Concatenation with `+` creates a new list without modifying either input.

Mutation changes an existing list object; rebinding a variable to a newly constructed list is a different operation.

```python
values = [1, 2, 3]
values[0] = 10
values[1:2] = [20, 21]
print(values)      # [10, 20, 21, 3]
combined = values + [4, 5]
print(combined)    # [10, 20, 21, 3, 4, 5]
print(values)      # [10, 20, 21, 3]
```

## Adding elements

`append(item)` adds one object to the end. `extend(iterable)` appends each item from an iterable. Appending a list creates one nested element, while extending with that list adds its elements separately. `insert(index, item)` inserts one object before a position.

These methods mutate the list and return `None`. Assigning their return value back to the list name would replace that binding with `None`.

```python
values = [1, 2]
returned = values.append([3, 4])
print(values)    # [1, 2, [3, 4]]
print(returned)  # None
values.extend([5, 6])
values.insert(0, 0)
print(values)    # [0, 1, 2, [3, 4], 5, 6]
```

## Removing elements

| Operation | Selection | Result |
| --- | --- | --- |
| `del values[index]` | Index | Deletes an element; `del` is a statement |
| `del values[start:stop]` | Slice | Deletes selected elements |
| `values.pop(index)` | Index, defaulting to the last element | Removes and returns the element |
| `values.remove(item)` | First equal value | Removes the first occurrence and returns `None` |
| `values.clear()` | All elements | Empties the list and returns `None` |

`remove()` raises `ValueError` when no equal value exists. `pop()` raises `IndexError` for an invalid index or an empty list. Removal by value searches for a match; removal near the beginning also shifts later elements. The cost depends on the list and removal position, so `remove()` is not inherently the wrong operation.

```python
values = [1, 2, 2, 3, 4]
values.remove(2)
print(values)       # [1, 2, 3, 4]
removed = values.pop(0)
print(removed)      # 1
print(values)       # [2, 3, 4]
del values[1:]
print(values)       # [2]
values.clear()
print(values)       # []
try:
    values.remove(99)
except ValueError:
    print("Value not present")
try:
    values.pop()
except IndexError:
    print("Empty list")
```

## Strings and lists

`list(text)` produces a list of characters. `text.split()` splits on runs of whitespace and drops leading and trailing whitespace. An explicit separator can contain several characters; repeated separators can produce empty fields. A nonempty string without that separator produces a one-element list. Empty text gives `[]` with `split()` but `['']` with an explicit separator.

`separator.join(parts)` combines strings with a separator between them. Every item must be a string; numeric items are not converted automatically. Splitting and joining are useful when processing fields from text data.

```python
text = "I <3 Python"
print(list("Hi"))                # ['H', 'i']
print(text.split())              # ['I', '<3', 'Python']
print(text.split("P"))           # ['I <3 ', 'ython']
print("  a   b  ".split())       # ['a', 'b']
print("a,,b".split(","))         # ['a', '', 'b']
print("abc".split(","))          # ['abc']
print("".split(), "".split(",")) # [] ['']
print(" ".join(["Hello", "Students"]))  # Hello Students
```

## Sorting and reversal

`values.sort()` sorts a list in place and returns `None`. `sorted(iterable)` returns a new sorted list. `reverse=True` selects descending order; `key` supplies a function whose results determine ordering.

Strings use lexicographic Unicode ordering by default, rather than locale-aware dictionary order. Arbitrary mixed types, such as integers and strings, cannot generally be ordered together without a suitable key.

`values.reverse()` reverses in place and returns `None`. `values[::-1]` creates a reversed shallow copy, and `reversed(values)` returns an iterator.

```python
numbers = [7, 0, 9, 1]
ordered = sorted(numbers)
print(ordered, numbers)  # [0, 1, 7, 9] [7, 0, 9, 1]
result = numbers.sort(reverse=True)
print(numbers, result)  # [9, 7, 1, 0] None
result = numbers.reverse()
print(numbers, result)  # [0, 1, 7, 9] None
print(numbers[::-1])    # [9, 7, 1, 0]
print(list(reversed(numbers)))  # [9, 7, 1, 0]
words = ["water", "alessio", "room", "python"]
print(sorted(words))          # ['alessio', 'python', 'room', 'water']
print(sorted(words, key=len)) # ['room', 'water', 'python', 'alessio']
```

## Aliasing and shallow copies

Assignment such as `alias = original` creates another reference to the same list. Mutation through either name is visible through both. `is` compares identity, while `==` compares values.

`original.copy()`, `original[:]`, and `list(original)` create new outer lists. They are shallow copies: element references are copied, so any nested mutable objects remain shared.

```python
original = [1, 2, 3]
alias = original
clone = original.copy()
print(alias is original, clone is original)  # True False
print(clone == original)                     # True
alias.append(4)
print(original)  # [1, 2, 3, 4]
print(clone)     # [1, 2, 3]

nested = [[1], [2]]
shallow = nested[:]
shallow[0].append(9)
print(nested)    # [[1, 9], [2]]
shallow.append([3])
print(len(nested), len(shallow))  # 2 3
```

## Independent nested structures

For an ordinary nested list, `copy.deepcopy()` copies nested mutable contents as well as the outer list. It is useful when edits to inner objects must be independent.

Repeating a list containing a mutable object repeats its reference. For example, `[[]] * 2` creates two references to one inner list. A comprehension can create a fresh inner list at each iteration.

```python
from copy import deepcopy

nested = [[1], [2]]
independent = deepcopy(nested)
independent[0].append(9)
print(nested, independent)  # [[1], [2]] [[1, 9], [2]]

shared_rows = [[]] * 2
shared_rows[0].append("x")
print(shared_rows)  # [['x'], ['x']]
separate_rows = [[] for _ in range(2)]
separate_rows[0].append("x")
print(separate_rows)  # [['x'], []]
```

## Removing items during traversal

A list iterator advances through positions in the underlying list. Deleting an element shifts later items left, so an item can be skipped when the iterator advances. Appending during iteration can also extend the traversal.

Building a filtered list avoids changing the traversed list. Iterating over a copy also permits removal from the original. Replacing values without changing length is a different case and does not cause this positional shift.

```python
values = [1, 2, 3, 4]
excluded = [1, 2, 5, 6]
for value in values:
    if value in excluded:
        values.remove(value)
print(values)  # [2, 3, 4]: 2 was skipped.

values = [1, 2, 3, 4]
filtered = [value for value in values if value not in excluded]
print(filtered)  # [3, 4]
print(values)    # [1, 2, 3, 4]

for value in values.copy():
    if value in excluded:
        values.remove(value)
print(values)    # [3, 4]
```
