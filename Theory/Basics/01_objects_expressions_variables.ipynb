# Objects, expressions, and variables

Python programs manipulate objects through expressions and statements. Objects have types, names refer to objects, and operators combine values.

## Computation and algorithms

Computation performs calculations and stores results. Declarative descriptions specify the result or relationship; imperative descriptions specify the steps used to obtain it.

An algorithm describes operations, their execution order, and a stopping condition. A program expresses these operations using statements, including assignments and definitions. Straight-line code executes in order; branches, loops, and function calls change that flow.

## Expressions and statements

An expression evaluates to a value with a type, such as `2 + 3`. A statement performs an action, such as binding a name with `=` or importing a module. A definition is a kind of statement, not a synonym for an expression.

A function call is an expression, even when used as a standalone statement. For example, `print()` displays output and returns `None`.

```python
total = 2 + 3
print(total)        # 5
print(type(total))  # <class 'int'>
```

## Interactive mode, scripts, and notebooks

Interactive mode evaluates complete commands as they are entered and displays expression results. A script stores code in a `.py` file; bare expressions in a script do not automatically appear as output.

A notebook stores code and notes in cells. Its running kernel retains names between cell executions, so execution order can differ from visual order. Starting from a fresh kernel and executing from top to bottom makes dependencies explicit.

Saving a script or notebook saves source code; it does not automatically preserve every object in the running session.

## Objects and types

Programs manipulate data objects, each with its own type.

A type determines which operations an object supports. Scalar objects represent individual values, while non-scalar objects have accessible internal structure, such as characters in a string or elements in a list.

| Type | Meaning | Example |
| --- | --- | --- |
| `int` | Integer, including zero and negative values | `-3` |
| `float` | Floating-point approximation of a real number | `2.0` |
| `bool` | Boolean value | `True`, `False` |
| `NoneType` | Type with the single value `None` | `None` |
| `str` | Sequence of Unicode characters | `"Python"` |

`type()` returns an object's type. `2` and `2.0` have equal numeric values but different types.

```python
print(type(-3))
print(type(2.0))
print(type(True))
print(type(None))
print(type("Python"))
print(2 == 2.0)  # True
```

## Arithmetic and precedence

Integer operations include addition, subtraction, multiplication, division, floor division, powers, remainder, and unary negation. In Python 3, `/` on integers returns a float, even when the quotient is a whole number.

| Operator | Meaning | Example |
| --- | --- | --- |
| `+`, `-`, `*` | Addition, subtraction, multiplication | `7 + 2` |
| `/` | True division | `7 / 2` gives `3.5` |
| `//` | Quotient rounded down toward negative infinity | `-7 // 2` gives `-4` |
| `%` | Remainder associated with floor division | `-7 % 2` gives `1` |
| `**` | Exponentiation | `2 ** 3` gives `8` |
| unary `-` | Negation | `-7` |

For integers and a nonzero divisor, `a == (a // b) * b + a % b`. Floor division differs from removing the fractional part when the quotient is negative.

Parentheses make grouping explicit. Multiplication and division bind more tightly than addition and subtraction. Exponentiation binds more tightly than unary minus on its left.

```python
print(7 / 2, 8 / 2)            # 3.5 4.0
print(7 // 2, -7 // 2)          # 3 -4
print(7 % 2, -7 % 2)            # 1 1
print((-7 // 2) * 2 + (-7 % 2)) # -7
print(7.0 // 2)                 # 3.0
print(2 + 3 * 4, (2 + 3) * 4)  # 14 20
print(-2 ** 2, (-2) ** 2)       # -4 4
```

## Floating-point representation

Just as one third has no finite decimal representation, many decimal fractions, including `0.1`, have no exact finite binary representation.

Python floats use finite-precision binary floating-point representation on ordinary platforms. They do not represent every real number exactly. The result of `0.1 + 0.2` is therefore slightly different from `0.3`.

`math.isclose()` compares values using tolerances. Its defaults suit this example; appropriate tolerances depend on the scale and meaning of the quantities.

```python
import math

approximate_sum = 0.1 + 0.2
print(approximate_sum)                   # 0.30000000000000004
print(approximate_sum == 0.3)            # False
print(math.isclose(approximate_sum, 0.3)) # True
```

## Conversion and rounding

`int()` converts a finite float to an integer by truncating toward zero, rather than rounding to the nearest integer.

`float()` converts suitable numbers or numeric strings to floats. `int()` accepts integer strings such as `"21"`, but decimal text such as `"2.5"` is not an integer literal. The names `int`, `float`, and `round` are lowercase and case-sensitive.

`round()` rounds a number. For halfway cases represented exactly, ties go to the even choice. Floating-point approximation can influence rounding of other decimal values.

```python
print(float(3), float("3.5"))  # 3.0 3.5
print(int(3.9), int(-3.9))     # 3 -3
print(int("21"))              # 21
print(round(2.5), round(3.5))  # 2 4
print(int(-7 / 2), -7 // 2)    # -3 -4
```

## Booleans and comparisons

`True` and `False` are capitalized Boolean values. Comparisons such as `==`, `!=`, `<`, `<=`, `>`, and `>=` produce Boolean results for the numbers used here.

Boolean operators combine simple tests: `and` requires both Boolean conditions, `or` requires at least one, and `not` flips the truth value.

| `a` | `b` | `a and b` | `a or b` | `not a` |
| --- | --- | --- | --- | --- |
| `True` | `True` | `True` | `True` | `False` |
| `True` | `False` | `False` | `True` | `False` |
| `False` | `True` | `False` | `True` | `True` |
| `False` | `False` | `False` | `False` | `True` |

`and` and `or` short-circuit: the right operand is evaluated only if needed. With non-Boolean operands, they return one of their operands rather than necessarily returning a `bool`.

```python
age = 21
has_ticket = True
print(age >= 18 and has_ticket)   # True
print(age < 18 or not has_ticket) # False
print(18 <= age < 30)             # True
print(age != 18)                  # True
print("" or "default")            # default
```

## `None`

`None` is an object representing the absence of a meaningful value. It is not the absence of an object itself. Its type is `NoneType`, and it is distinct from zero, an empty string, and `False`.

`is None` tests whether a value is this specific singleton object.

```python
result = None
print(result is None)   # True
print(result == 0)      # False
print(result == False)  # False
```

## Variables and naming

A variable gives a name to a value. Assignment with `=` binds a name to an object; `==` compares values. A later assignment can bind the same name to another object, including one of a different type.

Descriptive names express the role of a value and make code easier to change or share. Lowercase names with underscores, such as `circle_area`, are a common convention.

Ordinary ASCII names allow letters, digits, and underscores, but cannot start with a digit. Python also accepts many Unicode letters in identifiers. Spaces and punctuation such as hyphens are not allowed. Reserved keywords such as `True`, `False`, and `class` cannot be variable names. Names are case-sensitive: `true` is not the Boolean constant `True`.

```python
radius = 2
circle_area = math.pi * radius ** 2
original_area = circle_area
radius = 3
print(circle_area == original_area)  # True: changing radius does not update area.
circle_area = math.pi * radius ** 2
print(round(circle_area, 2))         # 28.27

value = 10
value = "ten"
print(value, type(value))
```

## Variables store results, not formulas

Naming values makes a calculation easy to reuse for different inputs. Assignment evaluates the expression on the right at that moment. Changing `radius` later does not automatically update a previously computed `circle_area`.

A name can be updated using its current value: `count = count + 1`. For numbers, `count += 1` expresses the same update concisely.

```python
count = 2
count = count + 1
count += 1
print(count)  # 4
```
