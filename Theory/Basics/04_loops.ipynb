# Loops

Iteration repeats a block of statements. A `while` loop repeats according to a condition; a `for` loop consumes items from an iterable. Both change control flow and can contain other loops or conditional blocks.

## Condition-based repetition: `while`

A `while` loop tests its condition before each iteration. Its body runs while the condition is truthy, so it can execute zero times. State used in the first test must already be initialized.

A counter controlling the loop is usually initialized before the loop and updated inside it. Resetting that counter on every iteration can prevent progress. A loop terminates when its condition becomes false or execution exits through a statement such as `break`.

```python
count = 0
while count < 3:
    print(count)
    count += 1
print("Final count:", count)  # 3

while count < 0:
    print("This block is skipped")
```

## Traversing an iterable: `for`

A `for` loop assigns each successive item from an iterable to its loop variable. It works with strings, lists, ranges, dictionaries, generators, and other iterables. The variable is not necessarily a numeric counter, and successive values need not increase.

A `for` loop ends when the iterable is exhausted, unless execution exits earlier. Its total iteration count need not be known in advance.

```python
for topic in ["strings", "branching", "loops"]:
    print(topic)

for character in "abc":
    print(character, end=" ")
print()
```

## Integer sequences with `range()`

`range(stop)` starts at zero with a step of one. `range(start, stop)` specifies a different start, and `range(start, stop, step)` also specifies the step. Arguments must be integers or support the integer index protocol; a zero step raises `ValueError`.

The stop value is always excluded. With a positive step, values stay below stop; with a negative step, they stay above stop. The last value is not generally `stop - 1`: it depends on the step. A range is empty when the starting value already lies outside these bounds.

A range represents an integer sequence without storing a list of all its values. `list()` materializes the values for display here.

```python
print(list(range(4)))         # [0, 1, 2, 3]
print(list(range(2, 5)))       # [2, 3, 4]
print(list(range(2, 8, 2)))    # [2, 4, 6]
print(list(range(5, 0, -2)))   # [5, 3, 1]
print(list(range(5, 0)))      # []
print(list(range(0)))         # []
```

## Counters and accumulators

A counter tracks progress or occurrences; an accumulator combines values across iterations. Both must retain their state between iterations, so initialization belongs before the loop that updates them.

For a fixed sequence of integers, `for` and `while` can express the same computation. `for` manages traversal, while the `while` version explicitly updates its counter.

```python
total_for = 0
for number in range(1, 5):
    total_for += number

total_while = 0
number = 1
while number < 5:
    total_while += number
    number += 1

print(total_for, total_while)  # 10 10
```

## Early termination: `break`

`break` immediately exits the innermost enclosing loop. Statements later in that loop body are skipped, and execution resumes after the loop. An unconditional `break` exits the first time execution reaches it, rather than necessarily after the first instruction.

A sentinel is a value marking the end of meaningful input. In this example, zero ends the sequence to be accumulated; later values are ignored.

```python
values = [4, 6, 0, 100]
total = 0
for value in values:
    if value == 0:
        break
    total += value
print(total)  # 10
```

## Skipping an iteration: `continue`

`continue` skips the remainder of the current iteration. A `for` loop proceeds to the next item; a `while` loop rechecks its condition.

In a `while` loop, progress must still be made on paths that use `continue`. An update placed after `continue` would be skipped and could leave the loop stuck.

```python
for character in "Luiss":
    if character == "i":
        continue
    print(character, end="")
print()  # Luss

number = 0
while number < 5:
    number += 1  # Updated before a possible continue.
    if number % 2 == 0:
        continue
    print(number, end=" ")
print()  # 1 3 5
```

## A statement that does nothing: `pass`

`pass` is a null statement, useful where Python requires a statement but no action is needed. It neither skips the remaining body nor exits the loop. This differs from both `continue` and `break`.

```python
for character in "Luiss":
    if character == "i":
        pass
    print(character, end="")
print()  # Luiss: the i is still printed.
```

## Nested loops

Loops can be nested. For each outer iteration, the inner loop starts again. A `break` in the inner loop leaves the outer loop running; ending both requires an additional decision at the outer level.

The example stops each inner traversal when the column reaches two, then stops the outer traversal when the row reaches one.

```python
for row in range(3):
    for column in range(4):
        if column == 2:
            break
        print(row, column)
    if row == 1:
        break
# Pairs: (0, 0), (0, 1), (1, 0), (1, 1)
```

## Normal completion and loop `else`

A loop can have an `else` block. It runs when a `for` loop exhausts its iterable or a `while` condition becomes false, including zero-iteration cases. It is skipped when that loop exits through `break`.

This is useful for searches: a match triggers `break`, while normal completion handles the absence of a match.

```python
target = 7
for value in [2, 4, 6]:
    if value == target:
        print("Found")
        break
else:
    print("Not found")
```

## Iteration beyond numeric sequences

Iteration also supports inspection and filtering. `dir(module)` returns attribute names, and a condition can select names of interest. An attribute name does not by itself imply that its value is a function.

The loop below selects names containing `log` from the standard-library `math` module.

```python
import math

for name in dir(math):
    if "log" in name:
        print(name)
```
