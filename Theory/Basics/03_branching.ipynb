# Branching

Branching changes the control flow of a program by choosing which statements execute according to a condition. Python uses `if`, `elif`, and `else` for conditional choices, and `match`/`case` for pattern matching. The pattern-matching examples require Python 3.10 or later.

## Conditions and blocks

An `if` statement evaluates the truth value of a condition. Its indented block runs when the condition is truthy; otherwise, the block is skipped. A colon ends the header, and indentation groups statements into the block. Four spaces per indentation level are conventional.

Statements at the surrounding indentation level execute after the conditional block, whether or not the condition was true.

```python
temperature = 32
if temperature > 30:
    print("High temperature")
    print("Above the threshold")
print("Measurement recorded")
```

## Two alternatives: `if` and `else`

An `else` block runs when the preceding `if` condition is false. Exactly one of the two blocks executes. `else` has no condition: it handles every case not covered by the `if`.

A remainder of zero after division by two identifies an even integer, including zero and negative even integers.

```python
number = -4
if number % 2 == 0:
    parity = "even"
else:
    parity = "odd"
print(parity)  # even
```

## Several alternatives: `elif`

An `if`/`elif`/`else` chain checks conditions in order. The first truthy condition selects its block; all remaining conditions and blocks in that chain are skipped. A final `else` handles the remaining cases. Without `else`, the chain can execute no block.

Order matters when conditions overlap. More specific conditions commonly precede broader ones.

```python
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "Below C"
print(grade)  # B
```

## Independent tests and exclusive choices

Separate `if` statements each evaluate their own condition, so several blocks can execute. An `elif` chain selects at most one block. The distinction is about the required behavior, rather than a general rule that one form is more efficient.

An `else` suits the entire remaining case. An `elif` suits a remaining case that still requires another test.

```python
number = 12
if number > 0:
    print("positive")
if number % 2 == 0:
    print("even")  # Both properties hold.

if number > 0:
    print("positive")
elif number % 2 == 0:
    print("even")  # Skipped because the first condition was true.
```

## Truth values and combined conditions

Conditions can be comparisons or other objects tested for truth. `False`, `None`, numeric zero, and empty strings or collections are falsy. Nonempty strings are truthy, including `"False"` and `"0"`.

`and`, `or`, and `not` combine conditions. Short-circuit evaluation can guard an operation: with `and`, a false left operand prevents evaluation of the right operand.

```python
print(bool(""), bool("False"), bool("0"))  # False True True

numerator = 10
denominator = 0
if denominator != 0 and numerator / denominator > 2:
    print("Quotient above two")
else:
    print("No valid quotient above two")  # Division by zero was not attempted.
```

## Nested conditions

A conditional can contain another conditional. The inner test is reached only when execution enters its enclosing block. Indentation determines which `if` each `else` belongs to.

Nested conditions can express a dependency between decisions. A flat `elif` chain often suits alternatives at the same level.

```python
number = 0
if number >= 0:
    if number == 0:
        sign = "zero"
    else:
        sign = "positive"
else:
    sign = "negative"
print(sign)  # zero
```

## Boundary cases and ties

Strict comparisons such as `>` exclude equality. A maximum-of-three calculation needs to account for ties; otherwise, equal largest values can leave every branch false. `>=` includes equality.

The built-in `max()` directly expresses the same operation when explicit branching is unnecessary.

```python
first, second, third = 7, 7, 3
if first >= second and first >= third:
    largest = first
elif second >= first and second >= third:
    largest = second
else:
    largest = third
print(largest)                     # 7
print(max(first, second, third))   # 7
```

## Conditional expressions

A conditional expression selects a value: `value_if_true if condition else value_if_false`. It is an expression, so it can appear on the right side of an assignment. A statement block is clearer when each branch requires several actions.

```python
number = 5
parity = "even" if number % 2 == 0 else "odd"
print(parity)  # odd
```

## Literal alternatives with `match` and `case`

Structural pattern matching is available from Python 3.10. `match` evaluates a subject and checks `case` patterns in order. The first matching case with a satisfied guard, if present, executes. There is no fall-through into later cases and no need for `break`.

Literal patterns can express a choice between specific values, similar to a switch statement. A final `case _` is a wildcard fallback. Without a matching case or fallback, no case body executes.

String literals are quoted. A bare name in a pattern normally captures the subject rather than comparing it with an existing variable. `_` matches without binding a name.

```python
first = 12
second = 3
operation = "/"

match operation:
    case "+":
        result = first + second
    case "-":
        result = first - second
    case "*":
        result = first * second
    case "/":
        if second == 0:
            result = "Division by zero"
        else:
            result = first / second
    case _:
        result = "Unknown operation"

print(result)  # 4.0
```

## Alternative patterns and guards

A `|` pattern matches any of its alternatives. A guard, written as `if condition` after a pattern, adds a Boolean condition. When a pattern matches but its guard is false, matching continues with later cases.

Pattern matching also supports structured data, beyond literal choices. For simple Boolean decisions, `if` and `elif` remain direct ways to express conditions.

```python
command = "save"
read_only = True

match command:
    case "save" | "write" if not read_only:
        message = "Write allowed"
    case "save" | "write":
        message = "Read-only mode"
    case "quit" | "exit":
        message = "Closing"
    case _:
        message = "Unknown command"

print(message)  # Read-only mode
```
