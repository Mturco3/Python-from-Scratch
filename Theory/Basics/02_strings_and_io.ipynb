# Strings and input/output

Strings represent text. Input supplies text to a program, explicit conversion interprets it as another type, and output formats values for display. Examples use fixed sample input so the notebook runs without waiting for keyboard input.

## Strings as objects

A Python `str` is an immutable sequence of Unicode characters. Letters, digits, spaces, and punctuation can all be part of a string; the text `"21"` differs from the integer `21`.

Single and double quotes delimit strings. Triple quotes also delimit strings and allow literal line breaks. The quote characters used as delimiters are not part of the resulting value.

````python
single = 'Python'
double = "Python"
multiline = """Python
notes"""
print(single == double)  # True
print(multiline)
print(type("21"), type(21))
````

## Concatenation and repetition

`+` concatenates strings, while `*` repeats a string an integer number of times. Direct concatenation of a string and a number is not supported; `str()` converts a value to text first.

These operations produce new strings. They do not modify the existing string objects.

````python
name = "Michele"
print("Hello, " + name)
print("ha" * 3)  # hahaha
print("Age: " + str(21))
print("Repeat: " + str(True))
print("ha" * 0)  # An empty string.
````

## Indexing and length

`len(text)` returns the number of Unicode code points in a string. Indexing is zero-based: `text[0]` selects the first character. Negative indices count from the end, with `text[-1]` selecting the last character. A selected character is itself a string of length one.

An index outside the string raises `IndexError`. An empty string has length zero and no valid character index. A displayed symbol can contain several code points, so string length does not always equal the number of visible symbols.

````python
text = "Hello"
print(len(text))        # 5
print(text[0], text[1], text[-1])  # H e o
print(len(""))          # 0
try:
    text[5]
except IndexError:
    print("Index outside the string")
````

## Slicing

`text[start:stop:step]` selects a substring. The start is included and the stop is excluded. The default step is one, and zero is invalid. With a positive step, omitted bounds span from the beginning to the end; with a negative step, the defaults span in reverse.

Negative bounds count from the end. `text[:-2]` excludes the last two characters, while `text[-2:]` selects them. `text[::-1]` reverses the code-point sequence.

Slices safely clip out-of-range bounds and can return an empty string. Unlike indexing, a slice need not select any existing character.

````python
text = "Hello"
print(text[1:4])   # ell
print(text[:-2])   # Hel
print(text[-2:])   # lo
print(text[::2])   # Hlo
print(text[::-1])  # olleH
print(text[:99])   # Hello
print(repr(text[4:1]))  # ''
print(repr(text[99:])) # ''
try:
    text[::0]
except ValueError:
    print("Slice step cannot be zero")
````

## Equality, membership, and iteration

`==` compares complete string values. `in` checks whether a substring occurs anywhere in a string, and `not in` checks its absence. These operations are case-sensitive. The empty string is a substring of every string.

A `for` loop traverses characters directly. A `while` loop can traverse the same string using an index and an explicit update.

````python
text = "Hello"
print(text == "Hello", text == "hello")  # True False
print("He" in text, "he" in text)        # True False
print("Me" not in text, "" in text)      # True True
for character in text:
    print(character, end=" ")
print()
index = 0
while index < len(text):
    print(text[index], end=" ")
    index += 1
print()
````

## Immutability and case conversion

String characters cannot be reassigned in place. A changed value is constructed from slices, concatenation, or a method result and can then be assigned to a name.

`capitalize()` titlecases the first character and lowercases the rest. `lower()` and `upper()` convert cased characters to lowercase and uppercase respectively. These methods return strings without modifying the original value.

````python
text = "hELLo"
print(text.capitalize())  # Hello
print(text.lower())       # hello
print(text.upper())       # HELLO
print(text)               # hELLo
replacement = "Y" + text[1:]
print(replacement)        # YELLo
try:
    text[0] = "Y"
except TypeError:
    print("Strings do not support item assignment")
````

## Searching and counting

| Method | Result |
| --- | --- |
| `text.count(sub)` | Number of non-overlapping occurrences |
| `text.startswith(sub)` | Whether the prefix matches |
| `text.endswith(sub)` | Whether the suffix matches |
| `text.find(sub)` | Lowest matching index, or `-1` if absent |
| `text.index(sub)` | Lowest matching index, or `ValueError` if absent |

These methods accept optional `start` and `end` bounds. The end is excluded. Search indices refer to the original string, even when the search is restricted to a region.

A `find()` result needs an explicit comparison with `-1`: index zero is falsy even though it means a match, while `-1` is truthy. Membership with `in` is simpler when only presence matters.

````python
text = "banana"
print(text.count("ana"))        # 1: overlapping matches are not counted
print(text.startswith("ban"))  # True
print(text.endswith("na"))     # True
print(text.find("na", 3))      # 4: an index into the original string
print(text.count("a", 1, 4))   # 2
print(text.find("ban"))        # 0
print(text.find("xyz"))        # -1
try:
    text.index("xyz")
except ValueError:
    print("Substring not found")
````

## Character classification

Classification methods return Boolean values without modifying the string.

| Method | Condition |
| --- | --- |
| `islower()` | At least one cased character, with all cased characters lowercase |
| `isupper()` | At least one cased character, with all cased characters uppercase |
| `istitle()` | At least one cased character, with uppercase/titlecase characters starting cased runs and lowercase characters following them |
| `isalpha()` | Nonempty, with every character a Unicode letter |
| `isalnum()` | Nonempty, with every character alphabetic or numeric |
| `isdecimal()` | Nonempty, with every character a Unicode decimal digit |
| `isdigit()` | Nonempty, with every character a digit, including some non-decimal digits |
| `isnumeric()` | Nonempty, with every character numeric, including some fractions |

Uncased characters such as digits and punctuation do not invalidate `islower()` or `isupper()`, but a string with no cased characters returns `False` for both.

Decimal characters form a subset of digit characters, which form a subset of numeric characters. These predicates describe characters, not complete numeric literals: a sign or decimal point makes the three numeric predicates false. Some characters accepted by `isdigit()` or `isnumeric()` cannot be parsed by `int()`.

````python
print("hello 123!".islower())  # True
print("HELLO 123!".isupper())  # True
print("123".islower())        # False
print("Hello World".istitle()) # True
print("Hello".isalpha(), "Hello2".isalpha())  # True False
print("Hello2".isalnum(), "Hello 2".isalnum()) # True False
print("".isalpha(), "".isdecimal())          # False False

superscript_two = chr(178)
one_half = chr(189)
print("2".isdecimal(), "2".isdigit(), "2".isnumeric())  # True True True
print(superscript_two.isdecimal(), superscript_two.isdigit(), superscript_two.isnumeric())  # False True True
print(one_half.isdecimal(), one_half.isdigit(), one_half.isnumeric())  # False False True
print("-2".isnumeric(), "2.5".isnumeric())  # False False
````

## Quotes and escape sequences

A backslash introduces an escape sequence. Escaped quotes represent characters inside a string rather than ending the string. A doubled backslash represents one literal backslash.

| Sequence | Meaning |
| --- | --- |
| `\\` | Literal backslash |
| `\'`, `\"` | Literal quote characters |
| `\n` | Newline |
| `\r` | Carriage return |
| `\t` | Horizontal tab |
| `\a` | Bell character |
| `\v` | Vertical tab |
| `\f` | Form feed |

Escape sequences control line changes, spacing, and other text behavior. Their visible effects depend on the terminal or renderer: a tab is not a fixed number of spaces, and a bell may produce no sound. `\v` is specifically a vertical tab, not a general newline.

`repr()` displays a representation that makes control characters visible as escapes.

````python
quotation = 'I said "What\'s going on?"'
path = "notes\\python"
print(quotation)
print(path)
print("first\nsecond")
print(repr("left\tright"))
print(repr("start\rend"))
print(repr("\a\v\f"))
````

## Output with `print()`

`print()` writes output. Multiple arguments are separated by commas in the call; those commas are Python syntax. A comma inside a quoted string is part of the text instead.

Several objects can be printed in one call without concatenating them manually. `print()` converts its arguments to text for display.

`sep` controls the text inserted between arguments and defaults to a space. `end` controls the text appended after the last argument and defaults to a newline. Both are keyword arguments to the function, not reserved Python keywords.

````python
print("apple", "banana")            # apple banana
print("apple, banana")               # The comma belongs to the string.
print("age", 21, True, sep=" | ")    # age | 21 | True
print("first", end=" -> ")
print("second")                     # first -> second
print("a", "b", sep="", end="!\n")  # ab!
````

## String formatting

Percent-style formatting uses placeholders: `%d` formats an integer representation, `%f` formats a floating-point representation, and `%s` formats text. `%f` defaults to six digits after the decimal point.

In `%8.2f`, `8` is the minimum total field width, including the sign and decimal point, and `2` is the number of digits after the decimal point. Width is not the number of digits before the decimal point. A value longer than the requested width is not cut off.

An f-string embeds expressions inside braces. A format such as `:.2f` displays two decimal places. Formatting controls the displayed text without changing the underlying numeric value.

````python
price = 12.3456
print("Name: %s; count: %d" % ("Michele", 3))
print("%f" % price)           # 12.345600
print(repr("%8.2f" % price))  # '   12.35'
print(f"Price: {price:.2f}")  # Price: 12.35
print(price)                 # 12.3456
````

## Input and explicit conversion

`input()` displays an optional prompt, waits for a line of input, and returns a string with the trailing newline removed. Digits typed at the keyboard are still text until converted.

```python
radius_text = input("Please insert radius value: ")
radius = float(radius_text)
```

The executable examples substitute fixed strings for the returned input. `int()` accepts integer text, while `float()` accepts suitable floating-point text. Invalid text raises `ValueError`; empty input is not automatically zero.

````python
integer_text = "21"  # A sample value returned by input().
decimal_text = "2.5"
print(integer_text * 2)       # 2121: string repetition
print(int(integer_text) * 2)  # 42: numeric multiplication
print(float(decimal_text))   # 2.5

try:
    int(decimal_text)
except ValueError:
    print("'2.5' is not valid integer text")
````

## Circle area: input, computation, output

A circle-area calculation combines text input, conversion to a number, computation with `math.pi`, and formatted output. The example uses `** 2` for the square and a fixed sample radius.

A nonnegative radius is assumed. The formatted output contains two decimal places; the stored area retains its floating-point value.

````python
import math

radius_text = "2.5"
radius = float(radius_text)
area = math.pi * radius ** 2
print(f"Area for radius {radius}: {area:.2f}")  # Area for radius 2.5: 19.63
````
