# Strings and input/output

Strings represent text. Input supplies text to a program, explicit conversion interprets it as another type, and output formats values for display. Examples use fixed sample input so the notebook runs without waiting for keyboard input.

## Strings as objects

A Python `str` is an immutable sequence of Unicode characters. Letters, digits, spaces, and punctuation can all be part of a string; the text `"21"` differs from the integer `21`.

Single and double quotes delimit strings. Triple quotes also delimit strings and allow literal line breaks. The quote characters used as delimiters are not part of the resulting value.

````python
single = 'Python'
double = "Python"
multiline = """Python
notes"""
print(single == double)  # True
print(multiline)
print(type("21"), type(21))
````

## Concatenation and repetition

`+` concatenates strings, while `*` repeats a string an integer number of times. Direct concatenation of a string and a number is not supported; `str()` converts a value to text first.

These operations produce new strings. They do not modify the existing string objects.

````python
name = "Michele"
print("Hello, " + name)
print("ha" * 3)  # hahaha
print("Age: " + str(21))
print("Repeat: " + str(True))
print("ha" * 0)  # An empty string.
````

## Quotes and escape sequences

A backslash introduces an escape sequence. Escaped quotes represent characters inside a string rather than ending the string. A doubled backslash represents one literal backslash.

| Sequence | Meaning |
| --- | --- |
| `\\` | Literal backslash |
| `\'`, `\"` | Literal quote characters |
| `\n` | Newline |
| `\r` | Carriage return |
| `\t` | Horizontal tab |
| `\a` | Bell character |
| `\v` | Vertical tab |
| `\f` | Form feed |

Escape sequences control line changes, spacing, and other text behavior. Their visible effects depend on the terminal or renderer: a tab is not a fixed number of spaces, and a bell may produce no sound. `\v` is specifically a vertical tab, not a general newline.

`repr()` displays a representation that makes control characters visible as escapes.

````python
quotation = 'I said "What\'s going on?"'
path = "notes\\python"
print(quotation)
print(path)
print("first\nsecond")
print(repr("left\tright"))
print(repr("start\rend"))
print(repr("\a\v\f"))
````

## Output with `print()`

`print()` writes output. Multiple arguments are separated by commas in the call; those commas are Python syntax. A comma inside a quoted string is part of the text instead.

Several objects can be printed in one call without concatenating them manually. `print()` converts its arguments to text for display.

`sep` controls the text inserted between arguments and defaults to a space. `end` controls the text appended after the last argument and defaults to a newline. Both are keyword arguments to the function, not reserved Python keywords.

````python
print("apple", "banana")            # apple banana
print("apple, banana")               # The comma belongs to the string.
print("age", 21, True, sep=" | ")    # age | 21 | True
print("first", end=" -> ")
print("second")                     # first -> second
print("a", "b", sep="", end="!\n")  # ab!
````

## String formatting

Percent-style formatting uses placeholders: `%d` formats an integer representation, `%f` formats a floating-point representation, and `%s` formats text. `%f` defaults to six digits after the decimal point.

In `%8.2f`, `8` is the minimum total field width, including the sign and decimal point, and `2` is the number of digits after the decimal point. Width is not the number of digits before the decimal point. A value longer than the requested width is not cut off.

An f-string embeds expressions inside braces. A format such as `:.2f` displays two decimal places. Formatting controls the displayed text without changing the underlying numeric value.

````python
price = 12.3456
print("Name: %s; count: %d" % ("Michele", 3))
print("%f" % price)           # 12.345600
print(repr("%8.2f" % price))  # '   12.35'
print(f"Price: {price:.2f}")  # Price: 12.35
print(price)                 # 12.3456
````

## Input and explicit conversion

`input()` displays an optional prompt, waits for a line of input, and returns a string with the trailing newline removed. Digits typed at the keyboard are still text until converted.

```python
radius_text = input("Please insert radius value: ")
radius = float(radius_text)
```

The executable examples substitute fixed strings for the returned input. `int()` accepts integer text, while `float()` accepts suitable floating-point text. Invalid text raises `ValueError`; empty input is not automatically zero.

````python
integer_text = "21"  # A sample value returned by input().
decimal_text = "2.5"
print(integer_text * 2)       # 2121: string repetition
print(int(integer_text) * 2)  # 42: numeric multiplication
print(float(decimal_text))   # 2.5

try:
    int(decimal_text)
except ValueError:
    print("'2.5' is not valid integer text")
````

## Circle area: input, computation, output

A circle-area calculation combines text input, conversion to a number, computation with `math.pi`, and formatted output. The example uses `** 2` for the square and a fixed sample radius.

A nonnegative radius is assumed. The formatted output contains two decimal places; the stored area retains its floating-point value.

````python
import math

radius_text = "2.5"
radius = float(radius_text)
area = math.pi * radius ** 2
print(f"Area for radius {radius}: {area:.2f}")  # Area for radius 2.5: 19.63
````
