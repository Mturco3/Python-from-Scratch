## Exercise 1: Circle

Create a Python class called `Circle` that models a geometric circle. The class should have the following features:

1. **Attributes**:
   - `radius`: A numeric attribute representing the radius of the circle.
   
2. **Methods**:
   - `__init__(self, radius)`: A constructor method to initialize the `radius` attribute when a `Circle` object is created.
   - `get_area(self)`: A method that returns the area of the circle (use the formula $\pi \times \text{radius}^2$).
   - `get_perimeter(self)`: A method that returns the perimeter (circumference) of the circle (use the formula $2 \times \pi \times \text{radius}$).
   - `display_info(self)`: A method that prints the radius, area, and perimeter of the circle.

```python
# Solution goes here
import math
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def get_area(self):
        return math.pi * self.radius * self.radius

    def get_perimeter(self):
        return 2 * math.pi * self.radius
```

### Test your code

Run this code to test your solution:

```python
# Test case 1: Circle with radius 1
circle1 = Circle(1)
assert abs(circle1.get_area() - math.pi) < 1e-6, "Test case 1 failed: Incorrect area"
assert abs(circle1.get_perimeter() - (2 * math.pi)) < 1e-6, "Test case 1 failed: Incorrect perimeter"

# Test case 2: Circle with radius 5
circle2 = Circle(5)
assert abs(circle2.get_area() - (math.pi * 25)) < 1e-6, "Test case 2 failed: Incorrect area"
assert abs(circle2.get_perimeter() - (2 * math.pi * 5)) < 1e-6, "Test case 2 failed: Incorrect perimeter"

# Test case 3: Circle with radius 0 (edge case)
circle3 = Circle(0)
assert abs(circle3.get_area() - 0) < 1e-6, "Test case 3 failed: Incorrect area"
assert abs(circle3.get_perimeter() - 0) < 1e-6, "Test case 3 failed: Incorrect perimeter"

# Test case 4: Circle with a floating-point radius
circle4 = Circle(2.5)
expected_area4 = math.pi * (2.5 ** 2)
expected_perimeter4 = 2 * math.pi * 2.5
assert abs(circle4.get_area() - expected_area4) < 1e-6, "Test case 4 failed: Incorrect area"
assert abs(circle4.get_perimeter() - expected_perimeter4) < 1e-6, "Test case 4 failed: Incorrect perimeter"
```

## Exercise 2: Bank Account

Create a Python class called `BankAccount` that represents a bank account. The class should have the following features:

1. **Attributes**:
   - `account_holder`: A string representing the name of the account holder.
   - `balance`: A numeric attribute representing the current balance in the account (initialized to 0 by default).




2. **Methods**:
   - `__init__(self, account_holder)`: A constructor method that initializes the `account_holder` and sets `balance` to 0.
   - `deposit(self, amount)`: Adds a specified amount to the balance. Print a message indicating the deposit amount and the new balance.
   - `withdraw(self, amount)`: Deducts a specified amount from the balance if sufficient funds are available. Print a message indicating the withdrawal amount and the new balance. If insufficient funds are available, print a warning message, without altering the balance.
   - `get_balance(self)`: Returns the current balance.
   - `display_info(self)`: returns, as string, a summary of the account holder’s name and current balance.

```python
# Solution goes here

class BankAccount:
    def __init__(self, account_holder, balance = 0):
        self.account_holder = account_holder
        self.balance = balance
    
    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive.")
            return
        self.balance += amount
        print(f"Deposited {amount}. New balance is {self.balance}.")

    def withdraw(self, amount):
        if amount > self.balance or amount <= 0:
            print("Invalid withdrawal amount.")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance is {self.balance}.")
    
    def get_balance(self):
        return self.balance
    
    def display_info(self):
        return f"Account holder: {self.account_holder}, Balance: {self.balance}"
```

### Test your code

Run this code to test your solution:

```python
# Test case 1: Create an account and check initial balance
account = BankAccount("Alice")
assert account.get_balance() == 0, "Test case 1 failed: Initial balance should be 0"

# Test case 2: Deposit a positive amount
account.deposit(100)
assert account.get_balance() == 100, "Test case 2 failed: Balance after deposit should be 100"

# Test case 3: Deposit a negative amount
account.deposit(-50)
assert account.get_balance() == 100, "Test case 3 failed: Balance should not change after negative deposit"

# Test case 4: Withdraw a valid amount
account.withdraw(50)
assert account.get_balance() == 50, "Test case 4 failed: Balance after withdrawal should be 50"

# Test case 5: Attempt to withdraw more than the balance
account.withdraw(100)
assert account.get_balance() == 50, "Test case 5 failed: Balance should not change after overdraft attempt"

# Test case 6: Withdraw a negative amount
account.withdraw(-20)
assert account.get_balance() == 50, "Test case 6 failed: Balance should not change after negative withdrawal"

# Test case 7: Display account information
info = account.display_info()
assert len(info) > 0, "Test case 7 failed: Incorrect display info"
```
