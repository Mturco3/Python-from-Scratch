## Sales Data Analysis
You are given a dataset containing information about sales transactions. Each row represents a sale, with the following columns:

| sale_id | product   | category   | quantity | price | date       | region   |
|---------|-----------|------------|----------|-------|------------|----------|
| 1       | Laptop    | Electronics| 2        | 900   | 2023-01-10 | North    |
| 2       | Headphone | Electronics| 5        | 50    | 2023-01-12 | South    |
| 3       | T-shirt   | Clothing   | 10       | 15    | 2023-01-13 | North    |
| 4       | Coffee    | Grocery    | 20       | 5     | 2023-01-14 | East     |
| 5       | Laptop    | Electronics| 1        | 950   | 2023-01-15 | West     |

**Columns explanation:**
- `sale_id`: Unique identifier for the sale (integer)
- `product`: Name of the product (string)
- `category`: Product category (string)
- `quantity`: Number of items sold (integer)
- `price`: Price per item (float)
- `date`: Date of sale (string, YYYY-MM-DD)
- `region`: Sales region (string)


### Exercise 1
Write a function `analyze_sales` that takes as input a DataFrame and a string `region_filter`, and returns a new DataFrame with the following properties:

1. Only include sales from the region specified by `region_filter`.
2. Add a new column `'total_value'` equal to `quantity * price`.
3. Only include sales where `'total_value'` is greater than 100.
4. Return only the columns: `sale_id`, `product`, `category`, `total_value`, and `date`.
5. The result should be sorted by `total_value` in descending order.


````python
# Write your function here
def analyze_sales(df, region_filter):
    df["total_value"] = df["quantity"] * df["price"]
    df = df[(df["region"] == region_filter) & (df["total_value"] > 100)]
    df.sort_values("total_value", ascending = False)
    return df[["sale_id", "product", "category", "total_value", "date"]]
````

### Test your code
Run this code to test your solution:

````python
import pandas as pd
from pandas.testing import assert_frame_equal

# Helper function to safely compare DataFrames ignoring dtype and small numeric differences
def assert_df_equal(df1, df2):
    """Safely compare two DataFrames ignoring dtype and small numeric differences."""
    assert_frame_equal(
        df1.reset_index(drop=True),
        df2.reset_index(drop=True),
        check_dtype=False,   # ignore int vs float
        check_exact=False,   # allow small rounding differences
        rtol=1e-5            # relative tolerance
    )

data = {
    'sale_id': [1, 2, 3, 4, 5],
    'product': ['Laptop', 'Headphone', 'T-shirt', 'Coffee', 'Laptop'],
    'category': ['Electronics', 'Electronics', 'Clothing', 'Grocery', 'Electronics'],
    'quantity': [2, 5, 10, 20, 1],
    'price': [900, 50, 15, 5, 950],
    'date': ['2023-01-10', '2023-01-12', '2023-01-13', '2023-01-14', '2023-01-15'],
    'region': ['North', 'South', 'North', 'East', 'West']
}
df = pd.DataFrame(data)

# Test 1: North region, two sales above threshold
expected1 = pd.DataFrame({
    'sale_id': [1, 3],
    'product': ['Laptop', 'T-shirt'],
    'category': ['Electronics', 'Clothing'],
    'total_value': [1800, 150],
    'date': ['2023-01-10', '2023-01-13']
})
assert_df_equal(analyze_sales(df, 'North'), expected1)

# Test 2: South region, one sale above threshold
expected2 = pd.DataFrame({
    'sale_id': [2],
    'product': ['Headphone'],
    'category': ['Electronics'],
    'total_value': [250],
    'date': ['2023-01-12']
})
assert_df_equal(analyze_sales(df, 'South'), expected2)

# Test 3: East region, no sale above threshold
expected3 = pd.DataFrame(columns=['sale_id', 'product', 'category', 'total_value', 'date'])
assert_df_equal(analyze_sales(df, 'East'), expected3)

# Test 4: West region, one sale above threshold
expected4 = pd.DataFrame({
    'sale_id': [5],
    'product': ['Laptop'],
    'category': ['Electronics'],
    'total_value': [950],
    'date': ['2023-01-15']
})
assert_df_equal(analyze_sales(df, 'West'), expected4)

print('All tests passed!')
````

## Exercise 2

Write a function `simple_sales_stats` that takes as input a DataFrame and a string `region_filter`, and returns a tuple with the following values:

1. The total number of sales in the specified region.
2. The total quantity of products sold in that region.
3. The mean price of products sold in that region.
4. The product (name) of the first sale with the highest price in that region.

**If there are no sales in the region, return `(0, 0, None, None)`.**

### Hint — read this only if you are not sure how to proceed

To find the product of the most expensive sale (i.e., the one with the highest price), you can use the `idxmax()` method, or follow these steps:

1. Compute the maximum value of the `price` column using the method `max()`.
2. Filter the DataFrame to include only the rows where the `price` equals this maximum value.
3. Retrieve the first element of the `product` column from the filtered DataFrame using `.iloc[0]`.

````python
# Write your function here
def simple_sales_stats(df, region_filter):
    if df[df["region"] == region_filter].shape[0] == 0:
        return (0,0, None, None)
    df = df[df["region"] == region_filter]
    df.sort_values("price", inplace=True, ascending= False)
    df.reset_index(inplace= True)
    total_sales = len(df)
    total_products = df["quantity"].sum()
    mean_price = df["price"].mean()
    highest_product = df.loc[0, "product"]
    return(total_sales, total_products, mean_price, highest_product)
````

Run this code to test your solution:

### Exercise 3
Write a function `filter_product_by_year` that takes as input a DataFrame, a string `product_name`, and an integer `year`, and returns a new DataFrame containing only the rows where:
- The `product` column matches `product_name` (exact match, case sensitive)
- The `date` column refers to a sale in the specified `year` (the year is the first 4 characters of the string)

The returned DataFrame should contain all columns and be sorted by `date` in ascending order.


````python
# Write your function here
def filter_product_by_year(df, product_name, year):
    df = df[(df["product"] == product_name) & (df["date"].str[:4] == str(year))]
    df.reset_index(inplace=True, drop=True)
    df.sort_values("date", inplace=True)
    return df
````

### Exercise 4

Write a function `assign_customer_codes(df, sale_codes, customer_map)` that takes as input:

1. `df`: A pandas DataFrame containing sales transactions, including the column `sale_id`.  
2. `sale_codes`: A list of strings in the format `'XXXX'` representing custom sale codes. These will be used as the new index of the DataFrame.  
3. `customer_map`: A dictionary where keys are sale codes (matching `sale_codes`) and values are customer codes in the format `'YYYY'`.

The function should:

1. Set the DataFrame’s index to the values from `sale_codes`.  
2. Remove the `sale_id` column (it is no longer needed).  
3. Add a new column called `customer_code`.  
4. Populate the `customer_code` column using the mapping in `customer_map`, where each key corresponds to a sale code (i.e., an index value).  

Finally, return the resulting DataFrame.

**Assumptions:**  

- `sale_codes` and `customer_map` have a **1:1 correspondence**; that is, every element in `sale_codes` is a key in `customer_map`, and every key in `customer_map` appears in `sale_codes`.

### Hint — read this only if you are not sure how to proceed

To handle the new column `customer_code`, you can initialize it with a default value (e.g., an empty string) and then populate it by assigning the correct values based on the DataFrame’s index.

---

#### Example

For the dataset:

| sale_id | product   | category   | quantity | price | date       | region   |
|---------|-----------|------------|---------|-------|------------|----------|
| 1       | Laptop    | Electronics| 2       | 900   | 2023-01-10 | North    |
| 2       | Headphone | Electronics| 5       | 50    | 2023-01-12 | South    |
| 3       | T-shirt   | Clothing   | 10      | 15    | 2023-01-13 | North    |
| 4       | Coffee    | Grocery    | 20      | 5     | 2023-01-14 | East     |
| 5       | Laptop    | Electronics| 1       | 950   | 2023-01-15 | West     |

```python
sale_codes = ['A1B2', 'C3D4', 'E5F6', 'G7H8', 'I9J0']
customer_map = {'A1B2': 'U111', 'C3D4': 'U222', 'E5F6': 'U333', 'G7H8': 'U444', 'I9J0': 'U555'}
assign_customer_codes(df, sale_codes, customer_map)
```

This should be the output:

|        | product   | category    | quantity | price | date       | region | customer_code |
|--------|-----------|-------------|----------|-------|------------|--------|---------------|
| A1B2   | Laptop    | Electronics | 2        | 900   | 2023-01-10 | North  | U111          |
| C3D4   | Headphone | Electronics | 5        | 50    | 2023-01-12 | South  | U222          |
| E5F6   | T-shirt   | Clothing    | 10       | 15    | 2023-01-13 | North  | U333          |
| G7H8   | Coffee    | Grocery     | 20       | 5     | 2023-01-14 | East   | U444          |
| I9J0   | Laptop    | Electronics | 1        | 950   | 2023-01-15 | West   | U555          |

````python
# Write your function here
def assign_customer_codes(df, sale_codes, customer_map):
    df["code"] = sale_codes
    df.set_index("code", inplace=True)
    df.drop("sale_id", axis = 1, inplace=True)
    idx = list(df.index)
    customer_codes = []
    for i in idx:
        customer_codes.append(customer_map[i])
    df["customer_code"] = customer_codes
    return df
````

````python
import pandas as pd

data = {
    'sale_id': [1, 2, 3, 4, 5],
    'product': ['Laptop', 'Headphone', 'T-shirt', 'Coffee', 'Laptop'],
    'category': ['Electronics', 'Electronics', 'Clothing', 'Grocery', 'Electronics'],
    'quantity': [2, 5, 10, 20, 1],
    'price': [900, 50, 15, 5, 950],
    'date': ['2023-01-10', '2023-01-12', '2023-01-13', '2023-01-14', '2023-01-15'],
    'region': ['North', 'South', 'North', 'East', 'West']
}
df = pd.DataFrame(data)
sale_codes = ['A1B2', 'C3D4', 'E5F6', 'G7H8', 'I9J0']
customer_map = {'A1B2': 'U111', 'C3D4': 'U222', 'E5F6': 'U333', 'G7H8': 'U444', 'I9J0': 'U555'}

result = assign_customer_codes(df, sale_codes, customer_map)

# Test 1: Index is set correctly
assert list(result.index) == sale_codes, "Test 1 failed"

# Test 2: 'customer_code' column is correct
assert result.loc['A1B2', 'customer_code'] == 'U111', "Test 2 failed"
assert result.loc['I9J0', 'customer_code'] == 'U555', "Test 2 failed"

# Test 3: 'sale_id' column is removed
assert 'sale_id' not in result.columns, "Test 3 failed"

print("All tests passed!")
````
