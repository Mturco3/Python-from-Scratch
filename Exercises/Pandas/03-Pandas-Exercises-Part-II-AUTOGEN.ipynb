## Exercise 1: Count Special Right Triangles
Write a function `count_special_triangles` that takes as input a list of tuples. Each tuple represents a right triangle and contains three elements:
- the first element is the base `a` (int or float),
- the second element is the height `b` (int or float),
- the third element is the color of the triangle (string).

The function should return the number of triangles that satisfy at least one of the two conditions:
- the color is `green` **and** the area is greater than `20`;
- the color is `red` **and** the area is not `10`.

If the list is empty or None, then the function needs to return `0`.

**Note**: Assume that all values for `a` and `b` are strictly greater than zero.
**Recall**: the area of a triangle with base `a` and height `b` is `(a * b)/2`.

**Example input:**
```python
triangles = [(3, 4, 'green'),(5, 10, 'blue'),(6, 4, 'red'),(2, 8, 'green')]

````python
# Write your function here
def count_special_triangles(triangles):
    result = 0
    if triangles == [] or triangles == None:
        return result
    for triangle in triangles:
        if triangle[2] == "green" and triangle[0]*triangle[1]/2 > 20:
            result += 1
        elif triangle[2] == "red" and triangle[0]*triangle[1]/2 != 10:
            result += 1
    return result
````

## Exercise 2: Cumulative Sum List

Write a function `cumulative_sum` that takes as input a list of integers and returns a **new list** where each element at index `i` represents the **sum of all elements** in the input list from index `0` up to and including index `i`.

If the input list is empty or `None`, the function should return an **empty list** `[]`.

### Example
If the input list is `[2, 4, 6, 8]`, then the output is `[2, 6, 12, 20]`, because:
- The first element is the sum of `[2]`
- The second element is the sum of `[2, 4]`
- The third element is the sum of `[2, 4, 6]`
- The fourth element is the sum of `[2, 4, 6, 8]`


````python
# Write your function here
def cumulative_sum(l):
    result = []
    if l == [] or l == None:
        return result
    cum_sum = 0
    for i in range(len(l)):
        cum_sum += l[i]
        result.append(cum_sum)
    return result
````

### Test your code
Run this code to test your solution:

````python
numbers = [2, 4, 6, 8]
assert cumulative_sum(numbers) == [2, 6, 12, 20]
````

````python
numbers2 = [1, 2, 3, 4, 5]
assert cumulative_sum(numbers2) == [1, 3, 6, 10, 15]
````

````python
numbers3 = [-1, -2, -3, -4]
assert cumulative_sum(numbers3) == [-1, -3, -6, -10]
````

````python
numbers4 = [5, -2, 3, -1, 4]
assert cumulative_sum(numbers4) == [5, 3, 6, 5, 9]
````

````python
numbers5 = [7]
assert cumulative_sum(numbers5) == [7]
````

````python
numbers6 = [1.5, 2.5, 3.0]
assert cumulative_sum(numbers6) == [1.5, 4.0, 7.0]
````

````python
assert cumulative_sum(None) == []
````

## Exercise 3: Filtering Special Drinks

Write a Python function `filter_special_drinks` that takes a DataFrame and a Boolean value as input and performs the following:

- If the Boolean is `True`:
    - Return **all drinks (full rows)** that are **carbonated** (`carbonated == True`) **and** **hot** (`temperature == 'hot'`).
- If the Boolean is `False`:
    - Return **all drinks (full rows)** where the **sugar** value is **less than the mean sugar** value in the DataFrame.

**Note:**
- You can assume the DataFrame does not contain any `NaN` values.
- The DataFrame always contains at least one row.

````python
# Write your function here
def filter_special_drinks(df, bool):
    mean_sugar = df["sugar"].mean()
    if bool:
        return df[(df["carbonated"] == True) & (df["temperature"] == "hot")]
    else:
        return df[df["sugar"] < mean_sugar]
````

## Exercise 4: Identifying Drinks with Significantly Increased Prices

Write a Python function `find_expensive_drinks` that takes as input a **DataFrame** and a **list of integers** representing updated prices for each drink.

The function should:

- Add the list `new_prices` as a new column in the DataFrame named `'new_price'`.
- Sort the values in the DataFrame by the `drink` column by **ascending order**.
- Compute the **mean of the original `price` column**.
- Return a **subset of the DataFrame** containing only the drinks whose `new_price` is **greater than double** the mean of the original prices, and **only the columns** `drink` and `new_price`.

**Note:**
- You can assume the DataFrame does not contain any `NaN` values.
- The DataFrame always contains at least one row.



````python
# Write your function here
def find_expensive_drinks(df, l):
    df["new_price"] = l
    df.sort_values("drink", ascending=True, inplace=True)
    df = df[df["new_price"] > 2*df["price"].mean()]
    return df[["drink", "new_price"]]
````
